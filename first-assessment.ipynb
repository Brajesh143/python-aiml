```python
m1 = int(input())
m2 = int(input())
m3 = int(input())
m4 = int(input())
m5 = int(input())

percentage = int((m1+m2+m3+m4+m5)/5);
if percentage >= 90:
    print(percentage)
    print("Grade A")
elif percentage >= 80 and percentage < 90:
    print(percentage)
    print("Grade B")
elif percentage >= 70 and percentage < 80:
    print(percentage)
    print("Grade C")
elif percentage >= 60 and percentage < 70:
    print(percentage)
    print("Grade D")
elif percentage >= 40 and percentage < 60:
    print(percentage)
    print("Grade E")
else:
    print(percentage)
    print("Grade F")
```

```python
def no_of_days(n):
    ans = 0
    if n == 2:
        ans = 28;
    elif n%2 != 0 or n == 8:
        ans = 31;
    else:
        ans = 30;
    return ans
no_of_days()
```

```python
def tax_to_pay(price):
    tax = 0;
    if price > 100000:
        tax = (price*20)/100;
    elif price > 75000 and price <= 100000:
        tax = (price*15)/100;
    elif price > 50000 and price <= 75000:
        tax = (price*10)/100;
    else:
        tax = price*5)/100;

    return round(tax, 1)
```
